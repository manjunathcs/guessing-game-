```
import random
lowest_number=1
highest_number=50
answer=random.randint(lowest_number,highest_number)
guesses=0
is_running=True

print("----------welcome to guessing game----------------")


while is_running:
   guess=input("enter the number between 1 and 50: ")


   if guess.isdigit():
      guess=int(guess)
      guesses +=1
      if guess<lowest_number or guess>highest_number:
           print("number is out of range ")

      elif guess < answer:
         print("number is TOO low !")
      elif guess > answer:
         print("number is Too high !")

      else:
         print("CORRECT!")
         print(f"total guess is {guesses} ")
         break


   else:
     print("enter the valid number")
```
